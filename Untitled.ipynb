```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. 生成示例数据（二维数据）
np.random.seed(42)  # 设置随机种子，确保结果可重复
mean = [0, 0]      # 均值
cov = [[1, 0.8], [0.8, 1]]  # 协方差矩阵（数据有相关性）
data = np.random.multivariate_normal(mean, cov, 100)  # 生成100个样本

# 2. 标准化数据（PCA通常需要标准化）
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data)

# 3. 使用PCA降维
pca = PCA(n_components=2)  # 保留2个主成分（这里为了可视化）
pca.fit(data_standardized)  # 拟合PCA模型
data_pca = pca.transform(data_standardized)  # 将数据转换到主成分空间

# 4. 绘制原始数据和PCA降维后的数据
plt.figure(figsize=(12, 5))

# 原始数据
plt.subplot(1, 2, 1)
plt.scatter(data[:, 0], data[:, 1], alpha=0.7)
plt.title("Original Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

# PCA降维后的数据
plt.subplot(1, 2, 2)
plt.scatter(data_pca[:, 0], data_pca[:, 1], alpha=0.7)
plt.title("PCA Transformed Data")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.tight_layout()
plt.show()

# 5. 输出PCA的主成分和解释方差
print("主成分（特征向量）：\n", pca.components_)
print("解释方差（每个主成分的贡献）：", pca.explained_variance_ratio_)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. 生成示例高维数据（假设有4个特征）
np.random.seed(42)
mean = np.zeros(4)  # 均值
cov = np.array([[1, 0.8, 0.6, 0.4],
                [0.8, 1, 0.7, 0.5],
                [0.6, 0.7, 1, 0.3],
                [0.4, 0.5, 0.3, 1]])  # 协方差矩阵
data = np.random.multivariate_normal(mean, cov, 100)  # 生成100个样本

# 2. 标准化数据
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data)

# 3. 使用PCA降维到二维
pca = PCA(n_components=2)
pca.fit(data_standardized)
data_pca = pca.transform(data_standardized)

# 4. 绘制降维后的数据和特征向量
plt.figure(figsize=(8, 6))
plt.scatter(data_pca[:, 0], data_pca[:, 1], alpha=0.7, label="Data Points")

# 绘制特征向量（主成分）
origin = np.zeros(2)  # 原点
for i, (component, variance) in enumerate(zip(pca.components_, pca.explained_variance_ratio_)):
    plt.quiver(*origin, *component[:2], scale=variance * 10, color=f"C{i+1}", label=f"PC{i+1} (Var: {variance:.2f})")

plt.title("PCA Transformed Data with Principal Components")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid()
plt.show()

# 5. 输出特征向量和解释方差
print("特征向量（主成分）：\n", pca.components_)
print("解释方差比例：", pca.explained_variance_ratio_)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_swiss_roll

# 生成瑞士卷数据
n_samples = 5000  # 样本数量
noise = 0.4       # 噪声强度（控制数据分布的扩散程度）
X, t = make_swiss_roll(n_samples=n_samples, noise=noise, random_state=42)

# X 是三维坐标数据，t 是沿瑞士卷流形的参数（颜色映射用）
print("数据形状:", X.shape)  # 输出: (1000, 3)
```

```python
from mpl_toolkits.mplot3d import Axes3D

# 创建3D画布
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# 绘制瑞士卷（用参数t作为颜色）
sc = ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=t, cmap=plt.cm.viridis, s=25)

# 设置坐标轴标签
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.title("Swiss Roll Dataset")
#plt.colorbar(sc, label="Parameter t")
plt.savefig("Swiss Roll Dataset.png")
plt.show()
```
